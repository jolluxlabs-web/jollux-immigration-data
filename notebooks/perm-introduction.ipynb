# PERM data: ingestion and recruitment intervals

This notebook creates a tiny synthetic CSV, loads it with the public `ingest_perm` API, and calculates a descriptive recruitment-start-to-filing interval.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

from jollux_immigration_data import ingest_perm, recruitment_statistics
```

All rows below are invented examples. The package accepts CSV or XLSX paths; using CSV keeps this example small and reproducible.

```python
perm_csv = """CASE_NUMBER,CASE_STATUS,RECEIVED_DATE,DECISION_DATE,EMP_BUSINESS_NAME,PWD_SOC_TITLE,JOB_TITLE,JOB_OPP_WAGE_FROM,JOB_OPP_WAGE_TO,JOB_OPP_WAGE_PER,PRIMARY_WORKSITE_CITY,PRIMARY_WORKSITE_STATE,RECR_INFO_JOB_START_DATE,RECR_INFO_JOB_END_DATE
A-EXAMPLE-1,Certified,2026-04-01,2026-08-01,Synthetic Analytics Incorporated,Data Scientists,Data Analyst,90000,110000,Year,Austin,TEXAS,2026-01-01,2026-01-31
A-EXAMPLE-2,Certified,2026-04-11,2026-08-11,Synthetic Systems LLC,Software Developers,Software Engineer,120000,,Year,Seattle,Washington,2026-02-01,2026-02-28
"""

with TemporaryDirectory() as temporary_directory:
    input_path = Path(temporary_directory) / "synthetic-perm.csv"
    input_path.write_text(perm_csv, encoding="utf-8")
    perm_records = list(ingest_perm(input_path))

len(perm_records)
```

```python
[(record["case_number"], record["employer_name"], record["worksite_state"], record["recruitment_start_date"], record["received_date"]) for record in perm_records]
```

## Recruitment-start-to-filing statistics

`recruitment_statistics` measures calendar days between `recruitment_start_date` and `received_date` for usable records. This is a derived descriptive interval, **not official PERM processing time**: it does not measure government adjudication or the time taken by the Department of Labor to process a case.

```python
recruitment_statistics(perm_records)
```
