# PWD data: a small end-to-end example

This notebook creates a tiny synthetic CSV, loads it with the public `ingest_pwd` API, and inspects normalized fields. It does not use Department of Labor data or make network requests.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

from jollux_immigration_data import ingest_pwd
```

The ingestion function expects the disclosure column names used below. The second row is intentionally excluded because its status is `Withdrawn`; the third is excluded because its visa class is not `PERM`.

```python
pwd_csv = """CASE_NUMBER,CASE_STATUS,RECEIVED_DATE,VISA_CLASS,EMPLOYER_LEGAL_BUSINESS_NAME,JOB_TITLE,PRIMARY_WORKSITE_CITY,PRIMARY_WORKSITE_STATE,PWD_SOC_TITLE,PWD_WAGE_RATE,PWD_UNIT_OF_PAY,PWD_OES_WAGE_LEVEL,PREVAIL_WAGE_DETERM_DATE,PWD_WAGE_EXPIRATION_DATE
P-EXAMPLE-1,Determination Issued,2026-01-10,PERM,Synthetic Analytics Inc.,Data Analyst,Austin,Texas,Data Scientists,90000,Year,Level 2,2026-02-01,2027-01-31
P-EXAMPLE-2,Withdrawn,2026-01-12,PERM,Synthetic Analytics Inc.,Engineer,Seattle,WA,Engineers,50,Hour,III,2026-02-02,2027-02-01
P-EXAMPLE-3,Determination Issued,2026-01-13,H-1B,Synthetic Systems LLC,Engineer,Boston,Massachusetts,Engineers,100000,Year,4,2026-02-03,2027-02-02
"""

with TemporaryDirectory() as temporary_directory:
    input_path = Path(temporary_directory) / "synthetic-pwd.csv"
    input_path.write_text(pwd_csv, encoding="utf-8")
    pwd_records = list(ingest_pwd(input_path))

len(pwd_records)
```

```python
[(record["case_number"], record["employer_name"], record["employer_normalized"], record["worksite_state"], record["wage_level"]) for record in pwd_records]
```

The normalized values are useful for matching and grouping, while the original employer name remains available for display and provenance.
